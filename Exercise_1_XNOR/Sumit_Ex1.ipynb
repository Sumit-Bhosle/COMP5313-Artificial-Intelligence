```python
import numpy as np
```

```python
# XNOR inputs
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

print(X)
```

```python
y = np.array([1,0,0,1])

print(y)
```

```python
# Equation -> z=w1x1 + w2x2 + b
```

```python
# Step Func

def step_func(z):
    if z >= 0:
        return 1
    else:
        return 0
        
```

```python
# Perceptron Func

def perceptron(x,weights,bias):
    z=np.dot(x,weights)+bias # z = w1x1 + w2x2 + b (dot product and addition)
    res = step_func(z)
    return res
```

```python
# Manual Test

x = np.array([1,0])
weights = np.array([1,1])
bias = -0.5

output = perceptron(x,weights,bias)

print("output :", output)
```

```python
# OR gate

or_weights = np.array([1,1])
or_bias = -0.5

for x in X:
    output = perceptron(x,or_weights,or_bias)
    print(x,"->",output)
```

```python
# AND gate

and_weights = np.array([1,1])
and_bias = -1.5

for x in X:
    output = perceptron(x,and_weights,and_bias)
    print(x,"->",output)
    
```

```python
# NAND gate

nand_weights = np.array([-1,-1])
nand_bias = 1.5

for x in X:
    output = perceptron(x,nand_weights,nand_bias)
    print(x,"->",output)
```

```python
# HIDDEN PERCEPTRON 1 : NAND

nand_output = []

for x in X:
    output = perceptron(x,nand_weights,nand_bias)
    nand_output.append(output)
nand_output = np.array(nand_output)

print("NAND OUTPUTS: ",nand_output)

    
```

```python
# HIDDEN PERCEPTRON 2 : OR
or_output = []

for x in X:
    output = perceptron(x,or_weights,or_bias)
    or_output.append(output)
or_output = np.array(or_output)

print("OR OUTPUTS: ",or_output)

    
```

```python
# Two Arrays into pair

hidden_outputs = np.column_stack((nand_output,or_output))
print (hidden_outputs)
```

```python
# Connecting Perceptron 3 

# XNOR using NAND + OR + final NAND

xnor_weights = np.array([-1, -1])
xnor_bias = 1.5

xnor_output = []

for x in hidden_outputs:
    output = perceptron(x, xnor_weights, xnor_bias)
    xnor_output.append(output)

xnor_output = np.array(xnor_output)

print("XNOR Output:", xnor_output)
```

```python
# Gradient Descent
## Differentiable Activation for training

def sigmoid(z):
    return 1/(1+np.exp(-z))
    
```

```python
# Calculate Sigmoid Prediction

weights = np.random.randn(2)
bias = np.random.randn()

z = np.dot(X,weights)+bias
predictions = sigmoid(z)

print("Weights: ",weights)
print("Bias: ", bias)
print("Predictions: ", predictions)
```

```python
# Calculate prediction error

error = predictions - y

print("Target: ",y)
print("Predictions: ", predictions)
print("Error: ",error)
```

```python
# Initialize weights and bias for the 2-layer XNOR network

w1 = np.random.randn(2,2)
b1 = np.random.randn(2)

w2 = np.random.randn(2)
b2 = np.random.randn()

print("w1 :",w1)
print("b1 :",b1)

print("w2 :",w2)
print("b2 :",b2)
```

```python
# Pass through XNOR 

# Hidden layer

z1 = np.dot(X,w1)+b1
hidden_output = sigmoid(z1)

# Output layer
z2 = np.dot(hidden_output,w2)+b2
output = sigmoid(z2)

print("Hidden Layer Output :")
print(hidden_output)
print("\nFinal Predictions :")
print(output)
```

```python
# Back Propogation 
# Calculate output layer error

output_error = output - y
output_delta = output_error*output*(1-output)

print("Output Error :")
print(output_error)

print("\nOutput Delta :")
print(output_delta)
```

```python
# Gradiant for output layer

dw2 = np.dot(hidden_output.T,output_delta)
db2 = np.sum(output_delta)

print("Gradiant for w2 :")
print(dw2)

print("\nGradiant for b2 :")
print(db2)

```

```python
# Calculate hidden layer Delta

hidden_error = output_delta[:,np.newaxis]*w2
hidden_delta = hidden_error*hidden_output*(1-hidden_output)

print("Hidden Error :")
print(hidden_error)

print("\nHidden Delta :")
print(hidden_delta)
```

```python
# Gradients for hidden layer

dw1 = np.dot(X.T, hidden_delta)
db1 = np.sum(hidden_delta, axis=0)

print("Gradient for w1:")
print(dw1)

print("\nGradient for b1:")
print(db1)
```

```python
# update Weights and Biases using gradient Descent

learning_rate = 0.1

w2 = w2 - learning_rate * dw2
b2 = b2 - learning_rate * db2

w1 = w1 - learning_rate * dw1
b1 = b1 - learning_rate * db1

print("Updated w1:")
print(w1)

print("\nUpdated b1:")
print(b1)

print("\nUpdated w2:")
print(w2)

print("\nUpdated b2:")
print(b2)
```

```python
# Automatic training using gradient descent

learning_rate = 1.0
epochs = 10000

loss_history = []

for epoch in range(epochs):

    # Forward pass
    z1 = np.dot(X, w1) + b1
    hidden_output = sigmoid(z1)

    z2 = np.dot(hidden_output, w2) + b2
    output = sigmoid(z2)

    # Output layer
    output_error = output - y
    output_delta = output_error * output * (1 - output)

    # Hidden layer
    hidden_error = output_delta[:, np.newaxis] * w2
    hidden_delta = hidden_error * hidden_output * (1 - hidden_output)

    # Gradients
    dw2 = np.dot(hidden_output.T, output_delta) / len(X)
    db2 = np.mean(output_delta)

    dw1 = np.dot(X.T, hidden_delta) / len(X)
    db1 = np.mean(hidden_delta, axis=0)

    # Update weights and biases
    w2 = w2 - learning_rate * dw2
    b2 = b2 - learning_rate * db2

    w1 = w1 - learning_rate * dw1
    b1 = b1 - learning_rate * db1

    # Mean squared error
    loss = np.mean((output - y) ** 2)
    loss_history.append(loss)

print("Training completed.")
print("Final loss:", loss_history[-1])
```

```python
# Test the trained XNOR network

z1 = np.dot(X, w1) + b1
hidden_output = sigmoid(z1)

z2 = np.dot(hidden_output, w2) + b2
final_output = sigmoid(z2)

predicted_output = (final_output >= 0.5).astype(int)

print("Target XNOR Output:")
print(y)

print("\nFinal Network Output:")
print(final_output)

print("\nPredicted XNOR Output:")
print(predicted_output)
```
